```python
# Comprobar coincidencias entre los diferentes conjuntos

pares = [
    ("Entrenamiento", "Validación"),
    ("Entrenamiento", "Prueba"),
    ("Validación", "Prueba")
]

for nombre_a, nombre_b in pares:

    df_a = datasets[nombre_a]
    df_b = datasets[nombre_b]

    ids_a = set(df_a["comment_id"])
    ids_b = set(df_b["comment_id"])

    textos_a = set(df_a["comment"].str.strip().str.lower())
    textos_b = set(df_b["comment"].str.strip().str.lower())

    coincidencias_ids = ids_a.intersection(ids_b)
    coincidencias_textos = textos_a.intersection(textos_b)

    print(f"\n{'='*55}")
    print(f"{nombre_a.upper()} VS {nombre_b.upper()}")
    print(f"{'='*55}")

    print("IDs coincidentes:", len(coincidencias_ids))
    print("Comentarios coincidentes:", len(coincidencias_textos))
```

```python
# Revisar comentarios repetidos entre entrenamiento y prueba

train = datasets["Entrenamiento"].copy()
test = datasets["Prueba"].copy()

train["texto_comparacion"] = train["comment"].str.strip().str.lower()
test["texto_comparacion"] = test["comment"].str.strip().str.lower()

revision_textos = train.merge(
    test,
    on="texto_comparacion",
    suffixes=("_train", "_test")
)

print("COMENTARIOS REPETIDOS:")
display(revision_textos[
    ["comment_id_train", "comment_id_test",
     "comment_train", "comment_test",
     "label_train", "label_test"]
])

print("\nCANTIDAD DE COINCIDENCIAS:", len(revision_textos))
```

```python
# Resumen de textos repetidos entre entrenamiento y prueba

resumen_textos = (
    revision_textos
    .groupby("texto_comparacion")
    .agg(
        Coincidencias=("texto_comparacion", "size"),
        Etiquetas_train=("label_train", lambda x: ", ".join(sorted(set(x)))),
        Etiquetas_test=("label_test", lambda x: ", ".join(sorted(set(x)))),
        IDs_train=("comment_id_train", lambda x: ", ".join(map(str, sorted(set(x))))),
        IDs_test=("comment_id_test", lambda x: ", ".join(map(str, sorted(set(x)))))
    )
    .reset_index()
)

print("TEXTOS ÚNICOS REPETIDOS:", len(resumen_textos))

print("\nTEXTOS CON DIFERENCIAS DE ETIQUETA:")
diferencias = resumen_textos[
    resumen_textos["Etiquetas_train"] != resumen_textos["Etiquetas_test"]
]

display(diferencias)

print("\nCANTIDAD DE TEXTOS CON DIFERENCIAS:", len(diferencias))
```

# NORMALIZACION

```python

import re
import unicodedata

def normalizar_texto(
    texto,
    quitar_tildes=False,
    quitar_puntuacion=False,
    quitar_numeros=False,
    quitar_emojis=False
):
    texto = str(texto).lower()

    # Sustituir URLs
    texto = re.sub(
        r"https?://\S+|www\.\S+",
        " url ",
        texto
    )

    # Sustituir menciones
    texto = re.sub(
        r"@\w+",
        " usuario ",
        texto
    )

    # Conservar el contenido de los hashtags
    texto = re.sub(r"#(\w+)", r"\1", texto)

    # Eliminar tildes si se solicita
    if quitar_tildes:
        texto = unicodedata.normalize("NFD", texto)
        texto = "".join(
            c for c in texto
            if unicodedata.category(c) != "Mn"
        )
        texto = unicodedata.normalize("NFC", texto)

    # Eliminar números si se solicita
    if quitar_numeros:
        texto = re.sub(r"\d+", " ", texto)

    # Eliminar puntuación si se solicita
    if quitar_puntuacion:
        texto = "".join(
            " " if unicodedata.category(c).startswith("P") else c
            for c in texto
        )

    # Eliminar emojis y símbolos si se solicita
    if quitar_emojis:
        texto = "".join(
            " " if unicodedata.category(c) in ("So", "Sk") else c
            for c in texto
        )

    # Normalizar espacios
    texto = re.sub(r"\s+", " ", texto).strip()

    return texto
```

```python

ejemplos = X_train.sample(n=5, random_state=42)

for comentario in ejemplos:
    print("ORIGINAL:")
    print(comentario)

    print("\nNORMALIZACIÓN COMPLETA:")
    print(normalizar_texto(comentario))

    print("\nSIN TILDES:")
    print(normalizar_texto(comentario, quitar_tildes=True))

    print("\nSIN PUNTUACIÓN:")
    print(normalizar_texto(comentario, quitar_puntuacion=True))

    print("\nSIN NÚMEROS:")
    print(normalizar_texto(comentario, quitar_numeros=True))

    print("\nSIN EMOJIS:")
    print(normalizar_texto(comentario, quitar_emojis=True))

    print("=" * 80)
```

```python
from nltk.corpus import stopwords

stopwords_es = set(stopwords.words('spanish'))

print("Total de stopwords en español:", len(stopwords_es))
print("Ejemplos:", list(sorted(stopwords_es))[:30])
```

```python
palabras_protegidas = {
    "no",
    "nunca",
    "jamas",
    "jamás",
    "ni",
    "sin",
    "tampoco",
    "nada",
    "nadie",
    "poco",
    "pero",
    "muy",
    "mas",
    "más"
}

stopwords_personalizadas = stopwords_es - palabras_protegidas

print("Stopwords originales:", len(stopwords_es))
print("Palabras protegidas:", len(palabras_protegidas))
print("Stopwords personalizadas:", len(stopwords_personalizadas))

print("\nPalabras protegidas:")
print(sorted(palabras_protegidas))
```

```python
for palabra in ["no", "nunca", "jamás", "ni", "sin", "tampoco", "nada"]:
    print(
        palabra,
        "->",
        "CONSERVADA" if palabra not in stopwords_personalizadas
        else "ELIMINADA"
    )
```

```python
def eliminar_stopwords(tokens):
    return [
        palabra
        for palabra in tokens
        if palabra not in stopwords_personalizadas
    ]
```

```python
prueba = ["no", "eres", "un", "idiota", "pero", "muy"]

print(eliminar_stopwords(prueba))
```

# LEMATIZACION

```python
def aplicar_stemming(tokens):
    return [
        stemmer.stem(palabra)
        if palabra.isalpha()
        else palabra
        for palabra in tokens
    ]
```

```python
palabras_protegidas_lematizacion = {
    "verle",
    "mierda",
    "jajaja",
    "jajajaja"
}

def aplicar_lematizacion(tokens):

    tokens_lematizados = []

    # Seleccionar las palabras que necesitan lematización
    palabras = [
        palabra for palabra in tokens
        if (
            palabra.isalpha()
            and not palabra.startswith("jaja")
            and palabra not in palabras_protegidas_lematizacion
        )
    ]

    # Procesar las palabras por lotes
    lemas = iter(nlp.pipe(palabras, batch_size=128))

    # Reconstruir la lista conservando el orden original
    for palabra in tokens:

        if (
            not palabra.isalpha()
            or palabra.startswith("jaja")
            or palabra in palabras_protegidas_lematizacion
        ):
            tokens_lematizados.append(palabra)

        else:
            lema = next(lemas)[0].lemma_
            tokens_lematizados.append(lema)

    return tokens_lematizados
```

```python
palabras_prueba = [
    "insultando",
    "insultos",
    "insultante",
    "jugando",
    "jugadores",
    "mejores",
    "bonitas",
    "estúpidos",
    "ofensivas"
]

print("PALABRA | STEMMING | LEMATIZACIÓN")
print("-" * 60)

for palabra in palabras_prueba:

    stemming = stemmer.stem(palabra)
    lema = nlp(palabra)[0].lemma_

    print(f"{palabra:15} | {stemming:15} | {lema}")
```

```python
# 1. Preprocesamiento sin reducción morfológica
def preprocesar_sin_reduccion(texto):

    tokens = tokenizar_texto(texto)
    tokens = eliminar_stopwords(tokens)

    return " ".join(tokens)


# 2. Preprocesamiento con stemming
def preprocesar_con_stemming(texto):

    tokens = tokenizar_texto(texto)
    tokens = eliminar_stopwords(tokens)
    tokens = aplicar_stemming(tokens)

    return " ".join(tokens)


# 3. Preprocesamiento con lematización
def preprocesar_con_lematizacion(texto):

    tokens = tokenizar_texto(texto)
    tokens = eliminar_stopwords(tokens)
    tokens = aplicar_lematizacion(tokens)

    return " ".join(tokens)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

def crear_vectorizador(ngramas=(1, 1)):

    return TfidfVectorizer(
        ngram_range=ngramas,
        min_df=2,
        max_features=50000,
        lowercase=False,
        token_pattern=r"(?u)\b\w+\b",
        sublinear_tf=False,
        use_idf=True,
        smooth_idf=True,
        norm="l2"
    )
```

```python
def seleccionar_preprocesamiento(texto, metodo="sin_reduccion"):

    if metodo == "sin_reduccion":
        return preprocesar_sin_reduccion(texto)

    elif metodo == "stemming":
        return preprocesar_con_stemming(texto)

    elif metodo == "lematizacion":
        return preprocesar_con_lematizacion(texto)

    else:
        raise ValueError("Método de preprocesamiento no válido")
```

```python
comentarios_prueba = [
    "No me gusta tu contenido",
    "Eres una persona maravillosa",
    "Eres un completo idiota",
    "No me gusta ese comentario",
    "Muchas gracias por tu apoyo"
]

metodos = [
    "sin_reduccion",
    "stemming",
    "lematizacion"
]

for metodo in metodos:

    print("\n" + "=" * 60)
    print("MÉTODO:", metodo.upper())

    textos_procesados = [
        seleccionar_preprocesamiento(texto, metodo)
        for texto in comentarios_prueba
    ]

    vectorizador = crear_vectorizador(ngramas=(1, 2))

    matriz = vectorizador.fit_transform(textos_procesados)

    print("Dimensiones de la matriz:", matriz.shape)
    print("Cantidad de características:", len(vectorizador.get_feature_names_out()))

    print("\nPrimeros términos del vocabulario:")
    print(vectorizador.get_feature_names_out()[:15])
```

# CREACION DE PIPELINE

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
```

```python
def crear_pipeline(metodo="sin_reduccion", ngramas=(1, 1), modelo=None):

    if modelo is None:
        raise ValueError("Debes proporcionar un modelo de clasificación")

    pipeline = Pipeline([
        (
            "preprocesamiento",
            FunctionTransformer(
                partial(preprocesar_lote, metodo=metodo),
                validate=False
            )
        ),
        (
            "tfidf",
            crear_vectorizador(ngramas=ngramas)
        ),
        (
            "clasificador",
            modelo
        )
    ])

    return pipeline
```

```python
from functools import partial

def preprocesar_lote(textos, metodo):
    return [
        seleccionar_preprocesamiento(texto, metodo)
        for texto in textos
    ]
```

```python
# Crear un pipeline de prueba
pipeline_prueba = crear_pipeline(
    metodo="sin_reduccion",
    ngramas=(1, 2),
    modelo=LogisticRegression(max_iter=1000)
)

# Separar las etapas de preprocesamiento y TF-IDF
pipeline_texto = pipeline_prueba[:-1]

# Transformar la muestra
matriz_prueba = pipeline_texto.fit_transform(muestra)

print("Dimensiones de la matriz:", matriz_prueba.shape)

print("\nPrimeras 15 características:")
print(
    pipeline_texto.named_steps["tfidf"]
    .get_feature_names_out()[:15]
)
```

```python
from sklearn.base import BaseEstimator, ClassifierMixin, clone
from sklearn.utils.class_weight import compute_sample_weight
import numpy as np

class XGBoostClasificador(BaseEstimator, ClassifierMixin):

    def __init__(self, modelo):
        self.modelo = modelo

    def fit(self, X, y):
        self.codificador_ = LabelEncoder()
        y_codificado = self.codificador_.fit_transform(y)

        pesos = compute_sample_weight(
            class_weight="balanced",
            y=y_codificado
        )

        self.modelo_ = clone(self.modelo)
        self.modelo_.fit(
            X,
            y_codificado,
            sample_weight=pesos
        )

        self.classes_ = self.codificador_.classes_
        return self

    def predict(self, X):
        predicciones = self.modelo_.predict(X)
        return self.codificador_.inverse_transform(
            np.asarray(predicciones, dtype=int)
        )

print("Adaptador de XGBoost preparado correctamente.")
```

```python
from sklearn.base import BaseEstimator, ClassifierMixin, clone, is_classifier
from sklearn.utils.class_weight import compute_sample_weight
import numpy as np

class XGBoostClasificador(ClassifierMixin, BaseEstimator):

    def __init__(self, modelo):
        self.modelo = modelo

    def fit(self, X, y):
        self.codificador_ = LabelEncoder()
        y_codificado = self.codificador_.fit_transform(y)

        pesos = compute_sample_weight(
            class_weight="balanced",
            y=y_codificado
        )

        self.modelo_ = clone(self.modelo)
        self.modelo_.fit(
            X,
            y_codificado,
            sample_weight=pesos
        )

        self.classes_ = self.codificador_.classes_
        return self

    def predict(self, X):
        predicciones = self.modelo_.predict(X)
        return self.codificador_.inverse_transform(
            np.asarray(predicciones, dtype=int)
        )

modelo_xgb_adaptado = XGBoostClasificador(modelo_xgb)

print("¿Se reconoce como clasificador?:", is_classifier(modelo_xgb_adaptado))
```

```python

import json
from pathlib import Path
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report
)

# Orden de las categorías
etiquetas = ['NO', 'NOE', 'OFG', 'OFP']


# Función para obtener los resultados de cada modelo
def obtener_resultados(y_real, y_pred):
    return {
        "accuracy": float(accuracy_score(y_real, y_pred)),

        "f1_macro": float(
            f1_score(y_real, y_pred, average='macro')
        ),

        "precision_macro": float(
            precision_score(
                y_real,
                y_pred,
                labels=etiquetas,
                average='macro',
                zero_division=0
            )
        ),

        "recall_macro": float(
            recall_score(
                y_real,
                y_pred,
                labels=etiquetas,
                average='macro',
                zero_division=0
            )
        ),

        "f1_por_clase": {
            categoria: float(valor)
            for categoria, valor in zip(
                etiquetas,
                f1_score(
                    y_real,
                    y_pred,
                    labels=etiquetas,
                    average=None
                )
            )
        },

        "matriz_confusion": confusion_matrix(
            y_real,
            y_pred,
            labels=etiquetas
        ).tolist(),

        "reporte": classification_report(
            y_real,
            y_pred,
            labels=etiquetas,
            target_names=etiquetas,
            output_dict=True,
            zero_division=0
        )
    }


# Construir el archivo con ambos modelos
resultados_exportar = {
    "conjunto": "test",
    "total_comentarios": int(len(y_test)),
    "categorias": etiquetas,
    "modelos": {
        "regresion_logistica": obtener_resultados(
            y_test, y_pred_lr
        ),

        "xgboost": obtener_resultados(
            y_test, y_pred_xgb
        )
    }
}


# Crear la carpeta de resultados
ruta_resultados = Path("resultados")
ruta_resultados.mkdir(exist_ok=True)


# Exportar a JSON
archivo_json = ruta_resultados / "resultados_modelos.json"

with open(archivo_json, "w", encoding="utf-8") as archivo:
    json.dump(
        resultados_exportar,
        archivo,
        ensure_ascii=False,
        indent=4
    )

print("Archivo generado correctamente:")
print(archivo_json.resolve())
```
